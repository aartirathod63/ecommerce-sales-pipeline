```python
orders["product_id"] = (
    orders["product_id"]
    .astype(str)
    .str.strip()
    .str.replace("P", "", regex=False)
)

print(orders["product_id"].head(10).tolist())
```

```python
orders["product_id"] = orders["product_id"].astype(int)

print(orders["product_id"].dtype)
print(products["product_id"].dtype)
```
